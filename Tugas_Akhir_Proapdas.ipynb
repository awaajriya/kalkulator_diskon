```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import datetime

# 1. Dekorasi Header
header = widgets.HTML(
    value="<h2 style='color: #2e7d32; text-align: center;'>🛒 Kalkulator Diskon Pro</h2>"
          "<p style='text-align: center;'>Hitung potongan harga dengan cepat dan simpan riwayatnya.</p>",
)

# 2. Komponen Input
style = {'description_width': 'initial'}
harga_input = widgets.IntText(description='Harga Barang (Rp):', value=0, style=style)
diskon_input = widgets.IntSlider(description='Diskon (%):', value=0, min=0, max=100, style=style)
nama_barang = widgets.Text(description='Nama Barang:', placeholder='Contoh: Sepatu', style=style)

# 3. Tombol & Output
hitung_btn = widgets.Button(description='Hitung & Simpan', button_style='success', icon='calculator', layout=widgets.Layout(width='98%', margin='10px 0px'))
reset_btn = widgets.Button(description='Reset Riwayat', button_style='danger', icon='trash', layout=widgets.Layout(width='98%'))
output_hasil = widgets.Output()
output_riwayat = widgets.Output()

riwayat_data = []

def hitung_semua(b):
    with output_hasil:
        clear_output()
        harga = harga_input.value
        diskon = diskon_input.value
        nama = nama_barang.value if nama_barang.value != "" else "Barang Tanpa Nama"

        if harga > 0:
            potongan = harga * (diskon / 100)
            total = harga - potongan
            waktu = datetime.datetime.now().strftime("%H:%M:%S")

            # Tampilan kartu hasil
            print(f"✅ Berhasil Dihitung!")
            print(f"{'='*30}")
            print(f"Produk      : {nama}")
            print(f"Hemat       : Rp {potongan:,.0f}")
            print(f"Total Bayar : Rp {total:,.0f}")
            print(f"{'='*30}")

            # Simpan ke riwayat
            riwayat_data.append(f"[{waktu}] {nama}: Rp {harga:,.0f} -> Rp {total:,.0f} (Disc {diskon}%)")
            update_riwayat()
        else:
            print("❌ Harga harus lebih dari 0!")

def update_riwayat():
    with output_riwayat:
        clear_output()
        if riwayat_data:
            print("📜 RIWAYAT PERHITUNGAN:")
            for item in reversed(riwayat_data): # Riwayat terbaru di atas
                print(item)

def reset_aplikasi(b):
    global riwayat_data
    riwayat_data = []
    with output_hasil: clear_output()
    with output_riwayat: clear_output()
    print("Riwayat telah dihapus.")

# Koneksi Fungsi ke Tombol
hitung_btn.on_click(hitung_semua)
reset_btn.on_click(reset_aplikasi)

# 4. Menyusun Layout Tampilan
input_panel = widgets.VBox([nama_barang, harga_input, diskon_input, hitung_btn, reset_btn])
tampilan_utama = widgets.HBox([
    widgets.VBox([widgets.Label("📝 INPUT DATA"), input_panel], layout=widgets.Layout(width='50%', border='1px solid #ddd', padding='10px')),
    widgets.VBox([widgets.Label("📊 HASIL & RIWAYAT"), output_hasil, output_riwayat], layout=widgets.Layout(width='50%', border='1px solid #ddd', padding='10px'))
])

# Jalankan Aplikasi
display(header, tampilan_utama)
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd

# Konfigurasi Halaman
st.set_page_config(page_title="Kalkulator Diskon Pro", page_icon="🛒")

# CSS Kustom untuk tampilan Mobile yang cantik
st.markdown("""
    <style>
    .main { background-color: #f5f7f9; }
    .stButton>button { width: 100%; border-radius: 10px; height: 3em; background-color: #2e7d32; color: white; }
    </style>
    """, unsafe_allow_html=True)

st.title("🛒 Kasir Diskon")
st.write("Hitung belanjaanmu jadi lebih mudah!")

# Input Area
with st.container():
    nama_barang = st.text_input("Nama Barang", placeholder="Misal: Kemeja")
    harga = st.number_input("Harga Barang (Rp)", min_value=0, step=1000, value=0)
    diskon = st.slider("Diskon (%)", 0, 100, 10)

# Tombol Hitung
if st.button("Hitung Sekarang"):
    if harga > 0:
        potongan = harga * (diskon / 100)
        total = harga - potongan

        st.balloons() # Efek perayaan

        # Kartu Hasil
        st.markdown(f"""
        <div style="background-color: white; padding: 20px; border-radius: 15px; border-left: 10px solid #2e7d32; box-shadow: 2px 2px 10px rgba(0,0,0,0.1);">
            <h3 style="margin:0;">Total Bayar:</h3>
            <h1 style="color: #2e7d32; margin:0;">Rp {total:,.0f}</h1>
            <p style="color: grey;">Hemat: Rp {potongan:,.0f}</p>
        </div>
        """, unsafe_allow_html=True)
    else:
        st.error("Masukkan harga barang terlebih dahulu!")
```
